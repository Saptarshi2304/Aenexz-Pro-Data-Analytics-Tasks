```python
# Import libraries
import pandas as pd

# 1. Load the Kaggle dataset
df = pd.read_csv("C:/Users/sapta/Downloads/titanic/train.csv")

# 2. Display first 5 rows
print("First 5 rows:")
print(df.head())

# 3. Display basic information
print("\nDataset Information:")
print(df.info())

# 4. Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Check for duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# 6. Remove duplicate rows
df = df.drop_duplicates()

# 7. Handle missing values

# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin has a very large number of missing values,
# so we remove this column
df = df.drop("Cabin", axis=1)

# 8. Check missing values again
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# 9. Check data types
print("\nData Types:")
print(df.dtypes)

# 10. Display cleaned dataset
print("\nCleaned Dataset:")
print(df.head())

# 11. Display final shape
print("\nFinal Dataset Shape:")
print(df.shape)

# 12. Save cleaned dataset
df.to_csv("cleaned_titanic.csv", index=False)

print("\nCleaned dataset saved as 'cleaned_titanic.csv'")
```
